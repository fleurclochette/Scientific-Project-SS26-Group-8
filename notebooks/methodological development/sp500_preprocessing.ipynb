# S&P 500 Data Preprocessing

This notebook prepares the S&P 500 daily price data for the methodological development analysis.

```python
# import libraries
import os # use it mainly for handling file paths
import numpy as np
import pandas as pd
```

```python
# File paths
raw_data_path = "data/raw/sp500_raw.csv"
processed_data_path = "data/processed/processed_returns.csv"
```

```python
# Load raw S&P 500 data
df = pd.read_csv(
    raw_data_path,
    parse_dates=["Date"]
)

df = df.sort_values("Date").reset_index(drop=True)

print(f"Observations: {len(df):,}")
print(f"Sample period: {df['Date'].min().date()} to {df['Date'].max().date()}")

df.head()
```

```python
# Validate required variables
required_columns = ["Date", "Close"]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

print("Missing values:")
print(df[required_columns].isna().sum())
```

```python
# Calculate daily logarithmic returns
df["LogReturn"] = (
    np.log(
        df["Close"] / df["Close"].shift(1)
    ) * 100
)

# Remove the first observation, for which no previous
# trading-day price is available
df = df.dropna(subset=["LogReturn"]).reset_index(drop=True)
```

```python
# Save processed dataset
df.to_csv(
    processed_data_path,
    index=False
)

print(f"Processed data saved to: {processed_data_path}")
```
