```python
# import libraries
import os   #use it mainly for handling file paths
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# define file path
# Your shared Google Drive project folder
project_dir = "/content/drive/MyDrive/Scientific Project 2026/SP500_Project_Data"

# Location of the raw dataset created in Notebook 01
raw_data_path = os.path.join(
    project_dir,
    "data",
    "raw",
    "sp500_raw.csv"
)

# Location where the processed dataset will be saved
processed_data_dir = os.path.join(
    project_dir,
    "data",
    "processed"
)

# Create the processed folder if it does not exist
os.makedirs(processed_data_dir, exist_ok=True)

processed_data_path = os.path.join(
    processed_data_dir,
    "processed_returns.csv"
)

# Check that the raw dataset exists
print("Checking raw dataset location...")

print("Raw file path:")
print(raw_data_path)

print("\nFile exists:", os.path.exists(raw_data_path))

# Load the raw S&P 500 datasets
print("\nLoading raw S&P 500 dataset...")

df = pd.read_csv(
    raw_data_path,
    index_col=0,
    parse_dates=True
)

print("Dataset loaded successfully!")

# Verify that the dataset has been imported correctly
print("\nFirst five observations:")
display(df.head())    # Shows the first five rows for inspection

print("\nDataset information:")
df.info()   # Displays information about the dataset

print("\nMissing values:")
print(df.isnull().sum())    # Checks for missing values in every column: If every column is zero, there are no missing values.

#fill missing values
df = df.bfill()   #Backfills missing values using the next available observation
df = df.ffill()   #Forward fills any remaining missing values using the previous observation.
print("\nMissing values after filling:")
print(df.isnull().sum())

# Calculate daily log return based on the closing price.
# Multiplication by 100 expresses the return as a percentage.

#calculate log returns
#Creates a new column called LogReturn.
df["LogReturn"] = (
    np.log(df["Close"] / df["Close"].shift(1))
) * 100

# The first observation has no previous closing price,
# so its log return is undefined and is removed.
df = df.dropna(subset=["LogReturn"])

# Save the processed dataset
df.to_csv(processed_data_path)

print("\n" + "-" * 60)
print("Data processing completed successfully!")
print(f"Processed dataset saved to:\n{processed_data_path}")
print("-" * 60)

# Display the processed dataset
print(f"\nTotal observations: {len(df)}")

display(df.head())

print("\nProcessed dataset information:")
df.info()
```
