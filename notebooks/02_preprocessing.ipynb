```
data/raw/raw_sp500.csv  #input #load the raw S&P 500 datqaset generated in 01 data collection
#import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
#Read the dataset exported from Notebook 01
df = pd.read_csv(
    "data/raw/raw_sp500.csv",
    index_col=0,
    parse_dates=True
)
#Verify that the dataset has been imported correctly
df.head()   #Shows the first five rows for inspection
df.info()   #Displays information about the dataset
df.isnull().sum()   #Checks for missing values in every column: If every column is zero, there are no missing values.
#fill missing values
df = df.bfill()   #Backfills missing values using the next available observation
df = df.ffill()   #Forward fills any remaining missing values using the previous observation.
#calculate log returns
#Creates a new column called LogReturn.
df['LogReturn'] = np.log(
    df['Close']/df['Close'].shift(1)
)*100
df = df.dropna()    #Removes the first row because it has no previous day's price, so its log return is undefined
df.to_csv("data/processed/processed_returns.csv")   #save the processed data set
```
